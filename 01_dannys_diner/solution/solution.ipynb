## Question 1
### What is the total amount each customer spent at the restaurant?

```python
# df = sales & menu merged
df = sales.merge(menu, on='product_id', how='left')
```

```python
sol_1 = df.groupby('customer_id').agg(total_spent=('price', 'sum')).reset_index()
```

## Question 2

### How many days has each customer visited the restaurant?
##### Restatement: No. of days a customer visited the restaurant
##### Output Grain: One record per customer

```python
result_2 = sales.groupby('customer_id').agg(visit_count = ('order_date', 'nunique')).reset_index()
print(result_2)
```

## Question 3

### What was the first item from the menu purchased by each customer?

#### Restatement: Customer's first purchase
#### Output Grain: One Record Per Customer of Item Purchased
#### First order ambiguous because multiple record for a customer could be on the same day

```python
result_3 = (
    df.sort_values(['customer_id','order_date'], ascending=True)
    .groupby('customer_id')['product_name']
    .first()
    .reset_index()
    )
print(result_3)
```

## Question 4

### What is the most purchased item on the menu and how many times was it purchased by all customers?

#### Restatement: Overall what item was purchased the most and how many times
#### Granularity: Product | Purchase_Count

```python
result_4 = (
    df.groupby('product_name')
    .agg(Product_Count=('order_date','count'))
    .sort_values(by='Product_Count', ascending=False)
    .head(1)
    .reset_index()
    )

print(result_4)
```
